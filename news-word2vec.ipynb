```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("suraj520/googlenews-vectors-negative300bingz-gz-format")

print("Path to dataset files:", path)
```

```python
import os

print(path)
print(os.listdir(path))
```

```python
model_path = path + "/GoogleNews-vectors-negative300.bin"

model = KeyedVectors.load_word2vec_format(model_path, binary=True)
```

```python
model['world']
```

```python
model.most_similar('cat')
```

```python
model.most_similar('criminals')
```

```python
model.most_similar('scam')
```

```python
model.similarity('man','woman')
```

```python
model.similarity('man','PHP')
```

```python
model.doesnt_match(['PHP','java','monkey'])
```

```python
vec = model['king'] - model['man'] + model['woman']
model.most_similar([vec])
```

```python
vec = model['INR'] - model ['India'] + model['England']
model.most_similar([vec])
```
