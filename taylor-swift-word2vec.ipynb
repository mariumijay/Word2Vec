```python
import nltk 
from nltk import sent_tokenize
from nltk.corpus import stopwords
from gensim.utils import simple_preprocess

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download("stopwords")
```

```python
stop_words = set(stopwords.words('english'))
print(list(stop_words)[:60])
```

```python
story = []

for filename in os.listdir(path):

    with open(os.path.join(path, filename), errors="ignore") as f:
        corpus = f.read()

    raw_sent = sent_tokenize(corpus)

    for sent in raw_sent:

        tokens = simple_preprocess(sent)

        # stopwords removal
        filtered = [word for word in tokens if word not in stop_words]

        # keep only alphabetic words
        clean_words = [w for w in filtered if w.isalpha()]

        story.append(clean_words)
        from collections import Counter

# flatten all words
flat_words = [word for sent in story for word in sent]

# frequency count
freq = Counter(flat_words)

# remove rare words
story = [[w for w in sent if freq[w] > 2] for sent in story]
```

```python
story
```

```python
model = gensim.models.Word2Vec(
    window=10,
    min_count=2
)
```

```python
model.build_vocab(story)
```

```python
model.train(story, total_examples=model.corpus_count, epochs=model.epochs)
```
